## Question

**Does evolvability peak at *intermediate* resource — and is the true control axis
`food_inc / tax` (gradient steepness) rather than `food_inc` alone?**

Intuition (checkpoint §B / §L): low resource → extinction / no standing variance → little
evolution; high resource → fitness *variance* collapses (drift-dominated) → little *adaptive*
evolution; intermediate → maximal adaptive evolution. The thermodynamic sharpening (§L,
Lotka / Odum maximum-power) predicts the productive peak tracks the **source/sink ratio
`food_inc / tax`** — the gradient the population can survive — not the absolute food input.

## Method (planned)

- **X-axis:** `food_inc / tax` (gradient steepness). Sweep `food_inc` across the extinction
  threshold up into the superabundant regime, at ≥2 `tax` levels, so the collapse onto the
  ratio can be tested (curves at different `tax` should *superimpose* when plotted vs the
  ratio).
- **Y-axes (two curves on one figure):** raw cumulative activity (expected roughly
  monotone / flat above the extinction threshold) **vs** excess activity
  (`dyn_excess` / `eg_excess`, and the canonical `A_new`) — expected **inverted-U** peaking in
  the interior. The *gap* between them is the whole point: raw activity stays high on the
  high-resource flat flank where excess collapses.
- **Instruments (Wave-2 dependencies):** the per-capita dissipation-rate readout (`metric-flux`
  / L1) supplies the principled `food_inc/tax` x-axis; `A_new` (`metric-anew` / G1) supplies the
  turnover-invariant excess. Scaffold this notebook now; drop figures in when those metrics
  land.
